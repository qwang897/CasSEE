```python
import os, sys
from pathlib import Path
_p = Path.cwd().resolve()
while not (_p / "src" / "pipeline" / "config.py").is_file():
    if _p.parent == _p:
        raise SystemExit("please run this notebook from inside the repository")
    _p = _p.parent
sys.path.insert(0, str(_p / "src"))

FOLD = 0  # 0-4, or None = all five folds

import numpy as np
import torch
from pipeline import config
from pipeline.test import casee_eval as ce
DEV = ce.device()
FOLDS = range(5) if FOLD is None else [FOLD]
```

```python
scores = {}
for k in FOLDS:
    out = ce.offtarget_scores(k, DEV)
    print("--- fold %d  (%d sets)" % (k, len(out)))
    for name in sorted(out):
        scores.setdefault(name, {})[k] = out[name]
        print("%-28s Pearson = %.4f" % (name, out[name]))
```

```python
print()
per_set = {n: float(np.mean(list(v.values()))) for n, v in scores.items()}
for name in sorted(per_set):
    v = list(scores[name].values())
    if len(v) > 1:
        print("%-28s mean = %.4f  std = %.4f" % (name, per_set[name], np.std(v)))
print()
print("%d sets, mean Pearson = %.4f" % (len(per_set), float(np.mean(list(per_set.values())))))
```
